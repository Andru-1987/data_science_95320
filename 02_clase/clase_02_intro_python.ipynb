```python
from datetime import datetime

# 1. Definimos una lista de diccionarios con distintos formatos
facturacion = [
    {
        "id": 1,
        "fecha": "2026-05-11",           # Formato ISO (YYYY-MM-DD)
        "monto": 3000.00
    },
    {
        "id": 2,
        "fecha": "11/05/2026",           # Formato Latino (DD/MM/YYYY)
        "monto": 1500.50
    },
    {
        "id": 3,
        "fecha": "May 11, 2026",         # Formato Anglo/Texto
        "monto": 4200.00
    },
    {
        "id": 4,
        "fecha": datetime.now().strftime("%d-%m-%Y %H:%M"), # Con hora incluída
        "monto": 950.00
    }
]
```

```python
facturacion[0]["monto"] + facturacion[1]["monto"] + facturacion[2]["monto"] + facturacion[3]["monto"]
```

```python
bolsa_de_plata = []
indice = 0

for factura in facturacion:
    bolsa_de_plata.append(factura["monto"])
    print(f"Esta bolsa esta creciendo -> {indice}")
    print(bolsa_de_plata)
    indice = indice + 1 # indice += 1

total = sum(bolsa_de_plata)
print(f"El gran total de esta recaudacion es:\t{total}" )
print(f"El promedio de esta recaudacion  es:\t{total / len(facturacion):.2f}" )
```

```python
def suma_de_facturas(facturacion):
    bolsa_de_plata = [  factura["monto"]  for factura in facturacion  ]
    print(bolsa_de_plata)
    return sum(bolsa_de_plata)
```

```python
suma_de_facturas(facturacion)
```

```python
# Ejemplo de acceso
for factura in facturacion:
    print(f"Factura #{factura['id']} - Fecha: {factura['fecha']} - Total: ${factura['monto']}")
```
