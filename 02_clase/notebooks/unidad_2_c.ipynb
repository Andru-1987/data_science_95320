```python
Dias =['Lunes','Martes','Miercoles','Jueves','Viernes','Sabado','Domingo']
Valores =[200, 225, 232, 221, 243, 256, 255]

import numpy as np

Dif = np.diff(Valores, n=1)

for x,y in zip(Dias[1:],Dif):
  if(x!= 'Lunes') & (y<0):
    print(x,y)
```
