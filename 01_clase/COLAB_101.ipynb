# Introducción a Google Colab


Google Colaboratory (Colab) es un entorno de cuaderno Jupyter basado en la nube que permite escribir y ejecutar código Python directamente en su navegador. Es especialmente útil para la ciencia de datos, el aprendizaje automático y la educación.

### Ventajas de Google Colab para la Ciencia de Datos:

1.  **Gratuito y accesible:** Colab es completamente gratuito y solo requiere una cuenta de Google. Esto lo hace accesible para cualquier persona interesada en la ciencia de datos, sin necesidad de invertir en hardware costoso o licencias de software.

2.  **Acceso a hardware potente (GPU/TPU):** Una de las mayores ventajas de Colab es que proporciona acceso gratuito a unidades de procesamiento gráfico (GPU) y unidades de procesamiento de tensores (TPU) de Google. Esto es crucial para ejecutar algoritmos de aprendizaje automático y redes neuronales que requieren una gran capacidad de cálculo.

3.  **Colaboración sencilla:** Colab permite la colaboración en tiempo real, similar a Google Docs. Múltiples usuarios pueden trabajar en el mismo cuaderno simultáneamente, lo que facilita el trabajo en equipo, la revisión de código y la enseñanza.

4.  **Entorno preconfigurado:** Viene con la mayoría de las bibliotecas de ciencia de datos y aprendizaje automático preinstaladas (como NumPy, Pandas, Scikit-learn, TensorFlow y PyTorch), lo que elimina la necesidad de configuraciones complejas y permite a los estudiantes empezar a codificar inmediatamente.

5.  **Integración con Google Drive:** Los cuadernos de Colab se guardan automáticamente en Google Drive, lo que facilita el almacenamiento, la organización y el acceso a sus proyectos desde cualquier lugar y dispositivo.

### ¿Cómo ayuda a los estudiantes?

Colab es una herramienta invaluable para los estudiantes porque:

*   **Democratiza el acceso a la tecnología:** Permite a los estudiantes aprender y experimentar con técnicas avanzadas de ciencia de datos y aprendizaje automático que de otra manera requerirían recursos económicos significativos.
*   **Fomenta el aprendizaje práctico:** Proporciona un entorno interactivo donde pueden escribir, ejecutar y probar código, lo cual es fundamental para el aprendizaje efectivo.
*   **Facilita proyectos grupales:** La capacidad de colaborar simplifica la realización de proyectos en equipo, una habilidad importante en el ámbito profesional.
*   **Reduce barreras técnicas:** Al encargarse de la infraestructura y las dependencias, los estudiantes pueden concentrarse en los conceptos y la resolución de problemas.

### Navegación Básica y Entorno de Google Colab

Los dos tipos principales de celdas en Google Colab son:

*   **Celdas de Código:** Estas celdas se utilizan para escribir y ejecutar código Python. Puedes escribir tus scripts, funciones o comandos y ejecutarlos directamente en el navegador.
*   **Celdas de Texto:** Estas celdas se utilizan para añadir documentación, explicaciones, títulos y otros textos formateados utilizando Markdown. Son esenciales para hacer tus cuadernos claros y comprensibles.

### Cómo Añadir Celdas

Existen varias formas de añadir nuevas celdas en Google Colab:

1.  **Botones de la barra de herramientas:** En la parte superior de la interfaz de Colab, verás dos botones principales:
    *   `+ Código`: Haz clic en este botón para añadir una nueva celda de código debajo de la celda actualmente seleccionada.
    *   `+ Texto`: Haz clic en este botón para añadir una nueva celda de texto debajo de la celda actualmente seleccionada.

2.  **Menú 'Insertar':** Puedes ir al menú `Insertar` en la parte superior de la ventana de Colab y seleccionar `Celda de código` o `Celda de texto` para añadir el tipo de celda deseado.

3.  **Flotante entre celdas:** Al pasar el ratón por el espacio entre dos celdas existentes, aparecerán pequeños iconos `+ Código` y `+ Texto`, lo que te permite insertar una celda en esa posición específica.

### Cómo Mover Celdas

Para reorganizar el orden de las celdas en tu cuaderno de Colab, puedes utilizar los siguientes métodos:

1.  **Flechas en la barra de herramientas de la celda:** Cuando seleccionas una celda (haciendo clic en ella), verás una barra de herramientas a la derecha o al pasar el ratón por encima. Dentro de esta barra, encontrarás flechas `↑` (arriba) y `↓` (abajo) que te permiten mover la celda seleccionada una posición hacia arriba o hacia abajo, respectivamente.

2.  **Menú 'Editar':** También puedes usar el menú superior `Editar`. Al hacer clic, verás opciones como `Mover celda hacia arriba` y `Mover celda hacia abajo`. Asegúrate de tener la celda que deseas mover seleccionada antes de usar estas opciones.

### Cómo Eliminar Celdas

Para eliminar una celda en tu cuaderno de Colab, puedes seguir estos pasos:

1.  **Icono de la papelera en la barra de herramientas de la celda:** Selecciona la celda que deseas eliminar. En la barra de herramientas que aparece a la derecha de la celda (o al pasar el ratón por encima), haz clic en el icono de la papelera (🗑️). Esto eliminará la celda inmediatamente.

2.  **Menú 'Editar':** Alternativamente, puedes ir al menú superior `Editar` y seleccionar `Eliminar celdas`. Asegúrate de que la celda a eliminar esté seleccionada antes de usar esta opción.

### El Panel de Archivos

El panel de archivos, ubicado en la barra lateral izquierda de Colab (icono de carpeta), te permite:

*   **Navegacion rapida:** Permite ver una estructura organizada de tu cuaderno, basada en los encabezados de Markdown (por ejemplo, `#`, `##`, `###`).
*   **Saltar a secciones:** Puedes hacer clic en cualquier entrada del indice para saltar directamente a esa seccion en tu cuaderno, lo que facilita la navegacion en cuadernos extensos.

Vamos a generar intencionadamente un error de código para ver cómo Colab lo maneja y qué información nos proporciona para depurar.

```python
print(variable_inexistente)
```

El error generado es un `NameError`. Esto significa que estamos intentando utilizar una variable (`variable_inexistente`) que no ha sido definida o asignada en nuestro código. ### Formato Markdown Básico

El Markdown es un lenguaje de marcado ligero que se utiliza para añadir formato a texto plano. Aquí te mostramos cómo usar algunos de los elementos más comunes en Colab:

#### Títulos y Subtítulos
Utiliza el símbolo `#` seguido de un espacio para crear títulos. Cuantos más `#` uses, menor será la jerarquía del título.

# Título Principal (con un solo '#')
## Título de Sección (con '##')
### Subsección (con '###')

#### Listas

Puedes crear listas ordenadas y no ordenadas:

**Listas no ordenadas (viñetas):**
*   Primer elemento
*   Segundo elemento
    *   Sub-elemento (con indentación)

**Listas ordenadas:**
1.  Primer paso
2.  Segundo paso
    1.  Sub-paso (con indentación)
3.  Tercer paso

#### Texto en Negrita e Itálica

Para resaltar texto, puedes usar asteriscos o guiones bajos:

*   Texto en **negrita** usando `**doble asterisco**`.
*   Texto en _itálica_ usando `_guiones bajos_`.
*   Texto en ***negrita e itálica*** usando `***triple asterisco***`.

## Conexión con Google Drive


### Por qué conectar Colab con Google Drive

Conectar Google Colab con Google Drive es fundamental por varias razones, especialmente para proyectos de ciencia de datos y aprendizaje automático:

1.  **Persistencia de datos:** Los entornos de Colab son temporales. Esto significa que, si no guardas tu trabajo, se perderá una vez que la sesión de Colab finalice. Al conectar Drive, puedes guardar tus cuadernos, datos, modelos entrenados y cualquier otro archivo generado directamente en tu almacenamiento en la nube, asegurando que tu progreso no se pierda.

2.  **Acceso a archivos grandes:** Google Drive es una excelente manera de almacenar y acceder a conjuntos de datos grandes que pueden ser demasiado voluminosos para subir directamente al entorno temporal de Colab. Una vez montado, Drive se comporta como un sistema de archivos local, permitiendo que tu código acceda a los datos almacenados en él.

3.  **Colaboración simplificada:** Si trabajas en equipo, puedes compartir carpetas y archivos en Google Drive, y todos los miembros del equipo podrán acceder a los mismos recursos a través de Colab, facilitando el trabajo colaborativo.

4.  **Organización y copia de seguridad:** Centraliza todos tus proyectos y recursos en un solo lugar en la nube, lo que facilita la organización y proporciona una copia de seguridad automática de tu trabajo.
